```python
# Keep only fishing observations inside SST coverage
fish_sst = training_fish[
    training_fish["CENTER_LAT"].between(
        float(sst.latitude.min()),
        float(sst.latitude.max())
    ) &
    training_fish["CENTER_LON"].between(
        float(sst.longitude.min()),
        float(sst.longitude.max())
    )
].copy()

print("Fishing observations to match:", len(fish_sst))


# Function to get nearest SST for a fishing observation
def get_nearest_sst(row):
    year = int(row["YEAR"])
    month = int(row["MONTH_START"])

    # SST dates belonging to same year and month
    matching_times = sst_times[
        (sst_times.year == year) &
        (sst_times.month == month)
    ]

    if len(matching_times) == 0:
        return np.nan

    # Pick SST date closest to the beginning of the fishing month
    target_date = pd.Timestamp(year=year, month=month, day=1)

    nearest_time = matching_times[
        np.argmin(np.abs(matching_times - target_date))
    ]

    # Extract nearest spatial grid cell
    value = sst["SST"].sel(
        time=nearest_time,
        latitude=row["CENTER_LAT"],
        longitude=row["CENTER_LON"],
        method="nearest"
    ).values

    return float(value)


# Apply to every fishing observation
fish_sst["SST"] = fish_sst.apply(get_nearest_sst, axis=1)

print("\nSST matching complete.")

print("\nMissing SST values:")
print(fish_sst["SST"].isna().sum())

print("\nDataset shape:")
print(fish_sst.shape)

print("\nSample:")
print(
    fish_sst[
        [
            "YEAR",
            "MONTH_START",
            "FISHING_GROUND_CODE",
            "CENTER_LAT",
            "CENTER_LON",
            "CPUE",
            "SST"
        ]
    ].head(10)
)
```

```python
# Check whether nearby SST cells can recover the missing observations

def nearby_sst_check(row, radius=1):
    year = int(row["YEAR"])
    month = int(row["MONTH_START"])

    matching_times = sst_times[
        (sst_times.year == year) &
        (sst_times.month == month)
    ]

    if len(matching_times) == 0:
        return np.nan

    target_date = pd.Timestamp(year=year, month=month, day=1)

    nearest_time = matching_times[
        np.argmin(np.abs(matching_times - target_date))
    ]

    # Find nearest grid indices
    lat_idx = int(
        np.abs(sst.latitude.values - row["CENTER_LAT"]).argmin()
    )
    lon_idx = int(
        np.abs(sst.longitude.values - row["CENTER_LON"]).argmin()
    )

    lat_start = max(0, lat_idx - radius)
    lat_end = min(len(sst.latitude), lat_idx + radius + 1)

    lon_start = max(0, lon_idx - radius)
    lon_end = min(len(sst.longitude), lon_idx + radius + 1)

    neighborhood = sst["SST"].sel(
        time=nearest_time
    ).isel(
        latitude=slice(lat_start, lat_end),
        longitude=slice(lon_start, lon_end)
    ).values

    valid = neighborhood[~np.isnan(neighborhood)]

    return len(valid)


missing_sst["NEARBY_VALID_CELLS"] = missing_sst.apply(
    nearby_sst_check,
    axis=1
)

print("Missing observations:", len(missing_sst))

print("\nNearby valid cells within 1 grid cell:")
print(
    missing_sst["NEARBY_VALID_CELLS"]
    .value_counts()
    .sort_index()
)

print(
    "\nRecoverable using immediate 3x3 neighborhood:",
    (missing_sst["NEARBY_VALID_CELLS"] > 0).sum()
)
```

```python
# Check whether missing SST cells have a valid SST
# on another date within the SAME month.

def count_valid_sst_in_month(row):
    year = int(row["YEAR"])
    month = int(row["MONTH_START"])

    matching_times = sst_times[
        (sst_times.year == year) &
        (sst_times.month == month)
    ]

    if len(matching_times) == 0:
        return 0

    # Find nearest spatial grid cell
    lat_idx = int(
        np.abs(sst.latitude.values - row["CENTER_LAT"]).argmin()
    )

    lon_idx = int(
        np.abs(sst.longitude.values - row["CENTER_LON"]).argmin()
    )

    values = sst["SST"].isel(
        time=[
            sst_times.get_loc(t)
            for t in matching_times
        ],
        latitude=lat_idx,
        longitude=lon_idx
    ).values

    return int(np.sum(~np.isnan(values)))


missing_sst["VALID_DATES_IN_MONTH"] = missing_sst.apply(
    count_valid_sst_in_month,
    axis=1
)

print("Missing observations:", len(missing_sst))

print("\nValid SST dates available within the same month:")
print(
    missing_sst["VALID_DATES_IN_MONTH"]
    .value_counts()
    .sort_index()
)

print(
    "\nPotentially recoverable:",
    (missing_sst["VALID_DATES_IN_MONTH"] > 0).sum()
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


def train_and_evaluate(features, name):

    categorical = ["GEAR"] if "GEAR" in features else []

    numerical = [
        f for f in features
        if f != "GEAR"
    ]

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "cat",
                OneHotEncoder(handle_unknown="ignore"),
                categorical
            ),
            (
                "num",
                "passthrough",
                numerical
            )
        ]
    )

    rf = RandomForestRegressor(
        n_estimators=300,
        max_depth=8,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    )

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", rf)
        ]
    )

    pipeline.fit(
        train_df[features],
        train_df["LOG_CPUE"]
    )

    predictions = pipeline.predict(
        test_df[features]
    )

    return {
        "Model": name,
        "MAE": mean_absolute_error(
            test_df["LOG_CPUE"],
            predictions
        ),
        "RMSE": np.sqrt(
            mean_squared_error(
                test_df["LOG_CPUE"],
                predictions
            )
        ),
        "R2": r2_score(
            test_df["LOG_CPUE"],
            predictions
        )
    }


# ------------------------------------------
# MODEL A: GEAR ONLY
# ------------------------------------------

model_a = train_and_evaluate(
    ["GEAR"],
    "A: Gear only"
)


# ------------------------------------------
# MODEL B: GEAR + SEASON
# ------------------------------------------

model_b = train_and_evaluate(
    [
        "GEAR",
        "MONTH_SIN",
        "MONTH_COS"
    ],
    "B: Gear + Season"
)


# ------------------------------------------
# MODEL C: ALL FEATURES
# ------------------------------------------

model_c = train_and_evaluate(
    [
        "GEAR",
        "MONTH_SIN",
        "MONTH_COS",
        "CENTER_LAT",
        "CENTER_LON",
        "SST"
    ],
    "C: Gear + Season + Location + SST"
)


# ------------------------------------------
# RESULTS
# ------------------------------------------

ablation_results = pd.DataFrame([
    model_a,
    model_b,
    model_c
])

print(
    ablation_results.to_string(index=False)
)
```

```python
# ==========================================
# TOP-K RANKING ACCURACY
# ==========================================

def top_k_overlap(results, k):
    actual_top = set(
        results.nlargest(k, "CPUE")[
            "FISHING_GROUND_CODE"
        ]
    )

    predicted_top = set(
        results.nlargest(k, "PRED_CPUE")[
            "FISHING_GROUND_CODE"
        ]
    )

    overlap = actual_top.intersection(predicted_top)

    return {
        "K": k,
        "Actual_Top_K": len(actual_top),
        "Predicted_Top_K": len(predicted_top),
        "Correct": len(overlap),
        "Overlap_%": 100 * len(overlap) / k
    }


top5 = top_k_overlap(results, 5)
top10 = top_k_overlap(results, 10)

print("TOP-K RANKING PERFORMANCE")
print("=========================")

print("\nTop-5:")
print(top5)

print("\nTop-10:")
print(top10)
```

```python
# ==========================================
# GROUND-LEVEL TOP-K EVALUATION BY MONTH
# ==========================================

def monthly_top_k_overlap(df, k):
    scores = []

    for (year, month), group in df.groupby(
        ["YEAR", "MONTH_START"]
    ):
        # Skip months with fewer than k grounds
        if len(group) < k:
            continue

        actual_top = set(
            group.nlargest(k, "ACTUAL_CPUE")[
                "FISHING_GROUND_CODE"
            ]
        )

        predicted_top = set(
            group.nlargest(k, "PRED_CPUE")[
                "FISHING_GROUND_CODE"
            ]
        )

        correct = len(
            actual_top.intersection(predicted_top)
        )

        scores.append({
            "YEAR": year,
            "MONTH": month,
            "GROUNDS": len(group),
            "CORRECT": correct,
            "OVERLAP_%": 100 * correct / k
        })

    return pd.DataFrame(scores)


top5_monthly = monthly_top_k_overlap(
    ground_month_results,
    5
)

print("MONTHLY TOP-5 RANKING")
print("=====================")

print(
    top5_monthly.to_string(index=False)
)

print("\nAverage Top-5 overlap:",
      round(top5_monthly["OVERLAP_%"].mean(), 2),
      "%")
```

```python
# ==========================================
# TOP-10 FISHING GROUND PREDICTOR
# ==========================================

def predict_top_grounds(year, month, gear):

    # Candidate fishing grounds from our historical data
    candidates = fish_geo_valid[
        fish_geo_valid["FISHING_GROUND_CODE"].notna() &
        fish_geo_valid["CENTER_LAT"].notna() &
        fish_geo_valid["CENTER_LON"].notna()
    ][
        [
            "FISHING_GROUND_CODE",
            "CENTER_LAT",
            "CENTER_LON"
        ]
    ].drop_duplicates(
        "FISHING_GROUND_CODE"
    ).copy()

    # Keep only grounds covered by our SST dataset
    candidates = candidates[
        candidates["CENTER_LAT"].between(
            float(sst.latitude.min()),
            float(sst.latitude.max())
        ) &
        candidates["CENTER_LON"].between(
            float(sst.longitude.min()),
            float(sst.longitude.max())
        )
    ].copy()

    # --------------------------------------
    # Find SST date for requested month
    # --------------------------------------

    matching_times = sst_times[
        (sst_times.year == year) &
        (sst_times.month == month)
    ]

    if len(matching_times) == 0:
        raise ValueError(
            f"No SST data available for {year}-{month:02d}"
        )

    target_date = pd.Timestamp(
        year=year,
        month=month,
        day=1
    )

    nearest_time = matching_times[
        np.argmin(
            np.abs(matching_times - target_date)
        )
    ]

    # --------------------------------------
    # Extract SST for each ground
    # --------------------------------------

    def extract_sst(row):

        value = sst["SST"].sel(
            time=nearest_time,
            latitude=row["CENTER_LAT"],
            longitude=row["CENTER_LON"],
            method="nearest"
        ).values

        return float(value)

    candidates["SST"] = candidates.apply(
        extract_sst,
        axis=1
    )

    # Remove grounds where satellite SST is unavailable
    candidates = candidates.dropna(
        subset=["SST"]
    ).copy()

    # --------------------------------------
    # Create seasonal features
    # --------------------------------------

    candidates["MONTH_SIN"] = np.sin(
        2 * np.pi * month / 12
    )

    candidates["MONTH_COS"] = np.cos(
        2 * np.pi * month / 12
    )

    candidates["GEAR"] = gear

    # --------------------------------------
    # Predict productivity
    # --------------------------------------

    candidates["PRED_LOG_CPUE"] = model.predict(
        candidates[feature_columns]
    )

    candidates["PRED_CPUE"] = np.exp(
        candidates["PRED_LOG_CPUE"]
    )

    # --------------------------------------
    # Rank
    # --------------------------------------

    top10 = candidates.sort_values(
        "PRED_CPUE",
        ascending=False
    ).head(10).reset_index(drop=True)

    top10["RANK"] = (
        top10.index + 1
    )

    return top10[
        [
            "RANK",
            "FISHING_GROUND_CODE",
            "CENTER_LAT",
            "CENTER_LON",
            "SST",
            "PRED_CPUE"
        ]
    ]
```

```python
# ==========================================
# GENERATE ALL PREDICTIONS FOR A MONTH
# ==========================================

def predict_all_grounds(year, month, gear):

    candidates = fish_geo_valid[
        fish_geo_valid["FISHING_GROUND_CODE"].notna() &
        fish_geo_valid["CENTER_LAT"].notna() &
        fish_geo_valid["CENTER_LON"].notna()
    ][
        [
            "FISHING_GROUND_CODE",
            "CENTER_LAT",
            "CENTER_LON"
        ]
    ].drop_duplicates(
        "FISHING_GROUND_CODE"
    ).copy()

    # SST coverage
    candidates = candidates[
        candidates["CENTER_LAT"].between(
            float(sst.latitude.min()),
            float(sst.latitude.max())
        ) &
        candidates["CENTER_LON"].between(
            float(sst.longitude.min()),
            float(sst.longitude.max())
        )
    ].copy()

    # Find SST dates in requested month
    matching_times = sst_times[
        (sst_times.year == year) &
        (sst_times.month == month)
    ]

    if len(matching_times) == 0:
        return pd.DataFrame()

    target_date = pd.Timestamp(
        year=year,
        month=month,
        day=1
    )

    nearest_time = matching_times[
        np.argmin(
            np.abs(matching_times - target_date)
        )
    ]

    # Extract SST
    def extract_sst(row):
        return float(
            sst["SST"].sel(
                time=nearest_time,
                latitude=row["CENTER_LAT"],
                longitude=row["CENTER_LON"],
                method="nearest"
            ).values
        )

    candidates["SST"] = candidates.apply(
        extract_sst,
        axis=1
    )

    candidates = candidates.dropna(
        subset=["SST"]
    ).copy()

    # Seasonal features
    candidates["MONTH_SIN"] = np.sin(
        2 * np.pi * month / 12
    )

    candidates["MONTH_COS"] = np.cos(
        2 * np.pi * month / 12
    )

    candidates["GEAR"] = gear

    # Predict
    candidates["PRED_LOG_CPUE"] = model.predict(
        candidates[feature_columns]
    )

    candidates["PRED_CPUE"] = np.exp(
        candidates["PRED_LOG_CPUE"]
    )

    # Rank ALL candidates
    candidates = candidates.sort_values(
        "PRED_CPUE",
        ascending=False
    ).reset_index(drop=True)

    candidates["PREDICTED_RANK"] = (
        candidates.index + 1
    )

    return candidates
```

```python
# ==========================================
# FULL GROUND-LEVEL TEST EVALUATION
# ==========================================

from scipy.stats import spearmanr

evaluation_rows = []

# Evaluate each gear separately
for gear in test_df["GEAR"].unique():

    gear_test = test_df[
        test_df["GEAR"] == gear
    ].copy()

    # Evaluate each available year/month
    for (year, month), group in gear_test.groupby(
        ["YEAR", "MONTH_START"]
    ):

        # Need at least 5 observed grounds
        if group["FISHING_GROUND_CODE"].nunique() < 5:
            continue

        # Actual ground-level CPUE
        actual = (
            group
            .groupby(
                "FISHING_GROUND_CODE",
                as_index=False
            )
            .agg(
                ACTUAL_CPUE=("CPUE", "mean")
            )
        )

        # Generate predictions for ALL candidate grounds
        predicted = predict_all_grounds(
            year,
            month,
            gear
        )

        # Merge actual observations with predictions
        merged = actual.merge(
            predicted[
                [
                    "FISHING_GROUND_CODE",
                    "PRED_CPUE",
                    "PREDICTED_RANK"
                ]
            ],
            on="FISHING_GROUND_CODE",
            how="left"
        )

        # Keep only grounds for which prediction exists
        merged = merged.dropna(
            subset=["PRED_CPUE"]
        )

        if len(merged) < 5:
            continue

        # Rank correlation
        rho, p_value = spearmanr(
            merged["ACTUAL_CPUE"],
            merged["PRED_CPUE"]
        )

        # Actual top 5
        actual_top5 = set(
            merged.nlargest(
                5,
                "ACTUAL_CPUE"
            )["FISHING_GROUND_CODE"]
        )

        # Predicted top 10 from the COMPLETE candidate pool
        predicted_top10 = set(
            predicted.head(10)[
                "FISHING_GROUND_CODE"
            ]
        )

        top5_hits = len(
            actual_top5.intersection(
                predicted_top10
            )
        )

        evaluation_rows.append({
            "YEAR": year,
            "MONTH": month,
            "GEAR": gear,
            "OBSERVED_GROUNDS": len(actual),
            "EVALUATED_GROUNDS": len(merged),
            "SPEARMAN": rho,
            "TOP5_IN_PRED_TOP10": top5_hits
        })


evaluation_df = pd.DataFrame(
    evaluation_rows
)

print("GROUND-LEVEL EVALUATION")
print("=======================")

print(
    evaluation_df.to_string(index=False)
)

print("\n--------------------------------")
print("SUMMARY")
print("--------------------------------")

if len(evaluation_df) > 0:

    print(
        "Evaluation periods:",
        len(evaluation_df)
    )

    print(
        "Average Spearman:",
        round(
            evaluation_df["SPEARMAN"].mean(),
            4
        )
    )

    print(
        "Median Spearman:",
        round(
            evaluation_df["SPEARMAN"].median(),
            4
        )
    )

    print(
        "Average actual Top-5 grounds found in predicted Top-10:",
        round(
            evaluation_df[
                "TOP5_IN_PRED_TOP10"
            ].mean(),
            2
        ),
        "/ 5"
    )

else:
    print(
        "No months had enough observations for evaluation."
    )
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# -----------------------------
# Time-based split
# -----------------------------
train = combined[combined["YEAR"] <= 2011].copy()
test = combined[combined["YEAR"].between(2012, 2014)].copy()

features = [
    "SST",
    "CENTER_LAT",
    "CENTER_LON",
    "MONTH_SIN",
    "MONTH_COS",
    "GEAR",
    "SPECIES_GROUP"
]

X_train = train[features]
y_train = train["LOG_CPUE"]

X_test = test[features]
y_test = test["LOG_CPUE"]

# -----------------------------
# Encode categorical features
# -----------------------------
categorical_features = ["GEAR", "SPECIES_GROUP"]
numeric_features = [
    "SST",
    "CENTER_LAT",
    "CENTER_LON",
    "MONTH_SIN",
    "MONTH_COS"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numeric",
            "passthrough",
            numeric_features
        )
    ]
)

# -----------------------------
# Random Forest
# -----------------------------
rf_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "model",
        RandomForestRegressor(
            n_estimators=400,
            max_depth=8,
            min_samples_leaf=3,
            random_state=42,
            n_jobs=-1
        )
    )
])

# Train
rf_model.fit(X_train, y_train)

# Predict
pred_log = rf_model.predict(X_test)

# Metrics
mae = mean_absolute_error(y_test, pred_log)
rmse = np.sqrt(mean_squared_error(y_test, pred_log))
r2 = r2_score(y_test, pred_log)

print("TRAINING ROWS:", len(train))
print("TEST ROWS:", len(test))

print("\nMODEL RESULTS")
print("----------------")
print("MAE :", round(mae, 4))
print("RMSE:", round(rmse, 4))
print("R²  :", round(r2, 4))
```

```python
def predict_top_10(year, month):
    """
    Predict and rank the 10 most productive fishing grounds
    for the requested year and month.
    """

    # Candidate grounds from our historical data
    candidates = combined[
        ["FISHING_GROUND_CODE", "CENTER_LAT", "CENTER_LON"]
    ].drop_duplicates("FISHING_GROUND_CODE").copy()

    # Keep grounds inside SST coverage
    candidates = candidates[
        candidates["CENTER_LAT"].between(2.5, 20.0) &
        candidates["CENTER_LON"].between(65.5, 95.5)
    ].copy()

    # Find SST for each candidate
    predictions = []

    for _, row in candidates.iterrows():

        month_data = ds.sel(
            time=ds.time.where(
                (ds.time.dt.year == year) &
                (ds.time.dt.month == month),
                drop=True
            )
        )

        if len(month_data.time) == 0:
            continue

        point = month_data["SST"].sel(
            latitude=row["CENTER_LAT"],
            longitude=row["CENTER_LON"],
            method="nearest"
        )

        sst_value = float(point.mean(skipna=True).values)

        if np.isnan(sst_value):
            continue

        # Predict for BOTH fisheries
        for species_group in ["TUNA", "BILLFISH"]:

            for gear in combined["GEAR"].unique():

                predictions.append({
                    "FISHING_GROUND_CODE":
                        row["FISHING_GROUND_CODE"],
                    "CENTER_LAT":
                        row["CENTER_LAT"],
                    "CENTER_LON":
                        row["CENTER_LON"],
                    "SST": sst_value,
                    "MONTH_SIN":
                        np.sin(2 * np.pi * month / 12),
                    "MONTH_COS":
                        np.cos(2 * np.pi * month / 12),
                    "GEAR": gear,
                    "SPECIES_GROUP":
                        species_group
                })

    prediction_df = pd.DataFrame(predictions)

    prediction_df["PRED_LOG_CPUE"] = rf_model.predict(
        prediction_df[features]
    )

    prediction_df["PRED_CPUE"] = np.exp(
        prediction_df["PRED_LOG_CPUE"]
    )

    # Average predictions across gear/species combinations
    top10 = (
        prediction_df
        .groupby(
            [
                "FISHING_GROUND_CODE",
                "CENTER_LAT",
                "CENTER_LON"
            ],
            as_index=False
        )
        .agg(
            PREDICTED_PRODUCTIVITY=("PRED_CPUE", "mean")
        )
        .sort_values(
            "PREDICTED_PRODUCTIVITY",
            ascending=False
        )
        .head(10)
        .reset_index(drop=True)
    )

    top10.insert(0, "RANK", range(1, len(top10) + 1))

    return top10
```
